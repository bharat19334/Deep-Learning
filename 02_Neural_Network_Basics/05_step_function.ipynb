```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def step_function(z):
    return (np.asarray(z) >= 0).astype(int)
```

```python
values = np.array([-3, -2, -1, 0, 1, 2, 3])
outputs = step_function(values)

for z, output in zip(values, outputs):
    print(f"z = {z:2d} , output = {output}")
```

```python
x = np.linspace(-5, 5, 500)
y = step_function(x)

plt.figure(figsize=(7, 4))
plt.step(x, y, where="post")
plt.xlabel("z")
plt.ylabel("Output")
plt.title("Step Activation Function")
plt.yticks([0, 1])
plt.grid(True)
plt.show()
```
