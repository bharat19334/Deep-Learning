```python
import numpy as np

# this is AND gate training data
X = np.array([
    [0, 0],
    [0, 1],
    [1, 0],
    [1, 1]
], dtype=float)

y = np.array([0, 0, 0, 1])

weights = np.zeros(2)
bias = 0.0
learning_rate = 0.01
epochs = 10
```

```python
def predict(x, weights, bias):
    z = np.dot(x, weights) + bias
    return 1 if z >= 0 else 0
```

```python
for epoch in range(epochs):
    errors = 0

    for x, target in zip(X, y):
        prediction = predict(x, weights, bias)
        error = target - prediction

        # Perceptron learning rule
        weights += learning_rate * error * x
        bias += learning_rate * error

        if error != 0:
            errors += 1

    print(
        f"Epoch {epoch + 1}: "
        f"errors={errors}, "
        f"weights={weights}, "
        f"bias={bias:.2f}"
    )

    if errors == 0:
        break
```

```python
print("\nFinal predictions:")
for x, target in zip(X, y):
    prediction = predict(x, weights, bias)
    print(f"Input: {x.astype(int)}, Target: {target}, Prediction: {prediction}")
```
