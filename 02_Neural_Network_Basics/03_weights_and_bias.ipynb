```python
import numpy as np
```

```python
x = np.array([1.0, 1.0])
weights = np.array([1.0, 1.0])
biases = [-0.5, -1.5, -2.5]
```

```python
for bias in biases:
    z = np.dot(x, weights) + bias
    output = 1 if z >= 0 else 0

    print(
        f"Weights: {weights}, "
        f"Bias: {bias}, "
        f"z: {z}, "
        f"Output: {output}"
    )
```
