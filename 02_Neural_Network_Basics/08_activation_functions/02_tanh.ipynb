```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def tanh(x):
    return np.tanh(x)
```

```python
x = np.linspace(-5, 5, 500)
y = tanh(x)

plt.plot(x, y)
plt.title("Tanh Activation")
plt.xlabel("x")
plt.ylabel("f(x)")
plt.grid(True)
plt.show()

print("Tanh(0):", tanh(0))
print("Tanh(2):", tanh(2))
print("Tanh(-2):", tanh(-2))
```
