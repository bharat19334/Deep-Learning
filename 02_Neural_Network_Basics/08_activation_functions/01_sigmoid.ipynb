```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def sigmoid(x):
    x = np.asarray(x, dtype=float)
    x = np.clip(x, -500, 500)

    return 1 / (1 + np.exp(-x))
```

```python
x = np.linspace(-10, 10, 500)
y = sigmoid(x)

plt.plot(x, y)
plt.title("Sigmoid Activation")
plt.xlabel("x")
plt.ylabel("f(x)")
plt.grid(True)
plt.show()

print("Sigmoid(0):", sigmoid(0))
print("Sigmoid(2):", sigmoid(2))
print("Sigmoid(-2):", sigmoid(-2))
```
