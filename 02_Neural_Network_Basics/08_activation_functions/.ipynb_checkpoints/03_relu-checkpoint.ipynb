```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def relu(x):
    return np.maximum(0, x)
```

```python
x = np.linspace(-5, 5, 500)
y = relu(x)

plt.plot(x, y)
plt.title("ReLU Activation")
plt.xlabel("x")
plt.ylabel("f(x)")
plt.grid(True)
plt.show()

print("ReLU:", relu(np.array([-3, -1, 0, 2, 4])))
```
