```python
import numpy as np

inputs = np.array([
    [0, 0],
    [0, 1],
    [1, 0],
    [1, 1]
], dtype=float)
```

```python
def step(z):
    return (z >= 0).astype(int)

def neuron(x, weights, bias):
    z = x @ weights + bias
    return z, step(z)
```

```python
# AND gate
and_weights = np.array([1.0, 1.0])
and_bias = -1.5

# OR gate
or_weights = np.array([1.0, 1.0])
or_bias = -0.5

z_and, output_and = neuron(inputs, and_weights, and_bias)
z_or, output_or = neuron(inputs, or_weights, or_bias)
```

```python
print("X1 X2 | AND | OR")
print("----------------")

for i, (x1, x2) in enumerate(inputs.astype(int)):
    print(f" {x1}  {x2}  |  {output_and[i]}  | {output_or[i]}")

assert np.array_equal(output_and, [0, 0, 0, 1])
assert np.array_equal(output_or, [0, 1, 1, 1])

print("\nBoth gates passed!")
```
