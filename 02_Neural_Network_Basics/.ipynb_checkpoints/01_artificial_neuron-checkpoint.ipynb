```python
import numpy as np
```

```python
x = np.array([1.0, 0.5])
w = np.array([0.8, 0.4])
b = -0.2

z = np.dot(x, w) + b

# this is Step activation
output = 1 if z >= 0 else 0
```

```python
print("Input:", x)
print("Weights:", w)
print("Bias:", b)
print("Weighted sum:", z)
print("Neuron output:", output)
```
