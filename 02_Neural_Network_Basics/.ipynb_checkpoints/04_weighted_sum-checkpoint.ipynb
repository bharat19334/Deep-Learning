```python
import numpy as np

X = np.array([
    [0, 0],
    [0, 1],
    [1, 0],
    [1, 1]
], dtype=float)

weights = np.array([0.7, 0.3])
bias = -0.4
```

```python
z = X @ weights + bias
```

```python
print("Input matrix:\n", X)
print("Weights:", weights)
print("Bias:", bias)
print("Weighted sums:", z)
```

```python
# Calculate one sample manually
x = X[3]

manual_z = x[0] * weights[0] + x[1] * weights[1] + bias

print("\nManual calculation for [1, 1]:", manual_z)
print("Matrix calculation:", z[3])

assert np.isclose(manual_z, z[3])
```
