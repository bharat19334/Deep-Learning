```python
import numpy as np

# A single perceptron
weights = np.array([0.8, 0.6])
bias = -0.7
```

```python
def predict(x):
    z = np.dot(x, weights) + bias
    return 1 if z >= 0 else 0
```

```python
samples = np.array([
    [0, 0],
    [0, 1],
    [1, 0],
    [1, 1]
])
```

```python
for x in samples:
    z = np.dot(x, weights) + bias
    output = predict(x)

    print(f"Input: {x}, z: {z:.2f}, Prediction: {output}")
```
