```python
import numpy as np
```

```python
def predict_one(x, weights, bias):
    z = np.dot(x, weights) + bias
    return 1 if z >= 0 else 0
```

```python
def fit(X, y, learning_rate=0.1, epochs=100):
    X = np.asarray(X, dtype=float)
    y = np.asarray(y, dtype=int)

    weights = np.zeros(X.shape[1])
    bias = 0.0

    for epoch in range(epochs):
        errors = 0

        for x, target in zip(X, y):
            prediction = predict_one(x, weights, bias)
            error = target - prediction

            weights += learning_rate * error * x
            bias += learning_rate * error

            if error != 0:
                errors += 1

        if errors == 0:
            print(f"Training converged at epoch {epoch + 1}")
            break

    return weights, bias
```

```python
def predict(X, weights, bias):
    X = np.asarray(X, dtype=float)

    return np.array([
        predict_one(x, weights, bias) for x in X
    ])
```

```python
# AND gate
X = np.array([
    [0, 0],
    [0, 1],
    [1, 0],
    [1, 1]
])
y = np.array([0, 0, 0, 1])
```

```python
weights, bias = fit(X, y, learning_rate=0.1, epochs=100)
predictions = predict(X, weights, bias)

print("\nLearned weights:", weights)
print("Learned bias:", bias)
print("Predictions:", predictions)
print("Expected:", y)

assert np.array_equal(predictions, y)
print("All test cases passed!")
```
