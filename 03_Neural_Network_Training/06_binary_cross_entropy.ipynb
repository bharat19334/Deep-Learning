```python
import numpy as np
```

```python
def binary_cross_entropy(y_true, probabilities):
    probabilities = np.clip(
        probabilities, 1e-9, 1 - 1e-9
    )

    loss = -(
        y_true * np.log(probabilities)
        + (1 - y_true) * np.log(1 - probabilities)
    )

    return np.mean(loss)
```

```python
y_true = np.array([1, 0, 1, 1])
probabilities = np.array([0.9, 0.2, 0.8, 0.6])

loss = binary_cross_entropy(y_true, probabilities)
```

```python
print("Actual values:", y_true)
print("Predicted probabilities:", probabilities)
print("Binary cross-entropy:", loss)
```
