```python
import numpy as np
```

```python
def sigmoid(z):
    return 1 / (1 + np.exp(-z))
```

```python
x = 2.0
w = 0.5
b = -0.1
target = 1.0
```

```python
# Forward pass
multiplication = x * w
z = multiplication + b
a = sigmoid(z)
loss = 0.5 * (a - target) ** 2
```

```python
# Backward pass
dL_da = a - target
da_dz = a * (1 - a)

dL_dz = dL_da * da_dz
dL_dw = dL_dz * x
dL_db = dL_dz
```

```python
print("x * w:", multiplication)
print("z:", z)
print("Activation:", a)
print("Loss:", loss)

print("\nGradients:")
print("dL/da:", dL_da)
print("da/dz:", da_dz)
print("dL/dw:", dL_dw)
print("dL/db:", dL_db)
```
