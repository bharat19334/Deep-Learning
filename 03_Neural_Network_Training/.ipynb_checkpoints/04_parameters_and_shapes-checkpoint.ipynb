```python
import numpy as np
```

```python
X = np.zeros((4, 2))

W1 = np.zeros((2, 3))
b1 = np.zeros((1, 3))

W2 = np.zeros((3, 1))
b2 = np.zeros((1, 1))

z1 = X @ W1 + b1
a1 = np.zeros_like(z1)

z2 = a1 @ W2 + b2
```

```python
print("Input shape:", X.shape)
print("W1 shape:", W1.shape)
print("b1 shape:", b1.shape)
print("Hidden output shape:", z1.shape)
print("W2 shape:", W2.shape)
print("b2 shape:", b2.shape)
print("Final output shape:", z2.shape)
```

```python
total_parameters = (
    W1.size + b1.size +
    W2.size + b2.size
)

print("\nTotal parameters:", total_parameters)
```
