```python
import numpy as np
```

```python
def sigmoid(z):
    return 1 / (1 + np.exp(-z))
def sigmoid_derivative_from_output(a):
    return a * (1 - a)
```

```python
# One sample
z = 0.8
y_true = 1.0

y_pred = sigmoid(z)
mse = 0.5 * (y_pred - y_true) ** 2
```

```python
# Chain rule for MSE
dL_da_mse = y_pred - y_true
da_dz = sigmoid_derivative_from_output(y_pred)
dL_dz_mse = dL_da_mse * da_dz
```

```python
# Binary cross-entropy with sigmoid output
bce = -(
    y_true * np.log(y_pred)
    + (1 - y_true) * np.log(1 - y_pred)
)
```

```python
# Combined derivative for BCE + sigmoid
dL_dz_bce = y_pred - y_true

print("Prediction:", y_pred)
print("MSE loss:", mse)
print("dL/da for MSE:", dL_da_mse)
print("da/dz:", da_dz)
print("dL/dz for MSE:", dL_dz_mse)
print("BCE loss:", bce)
print("dL/dz for BCE:", dL_dz_bce)
```
