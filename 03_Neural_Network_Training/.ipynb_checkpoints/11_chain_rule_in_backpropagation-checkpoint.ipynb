```python
import numpy as np

def sigmoid(z):
    return 1 / (1 + np.exp(-z))
```

```python


# Given values
x = 2.0
w = 0.5
b = 0.1
y_true = 1.0
```

```python
# Forward calculations
z = x * w + b
a = sigmoid(z)
loss = 0.5 * (a - y_true) ** 2
```

```python
# Derivative of loss with respect to activation
dL_da = a - y_true

# Derivative of activation with respect to z
da_dz = a * (1 - a)

# Derivative of z with respect to weight
dz_dw = x

# Chain rule
dL_dw = dL_da * da_dz * dz_dw
```

```python
# Numerical gradient
epsilon = 1e-5

def calculate_loss(weight):
    prediction = sigmoid(x * weight + b)
    return 0.5 * (prediction - y_true) ** 2
```

```python
numerical_gradient = (
    calculate_loss(w + epsilon)
    - calculate_loss(w - epsilon)
) / (2 * epsilon)
```

```python
print("Loss:", loss)
print("dL/da:", dL_da)
print("da/dz:", da_dz)
print("dz/dw:", dz_dw)
print("Analytical gradient:", dL_dw)
print("Numerical gradient:", numerical_gradient)

assert np.isclose(
    dL_dw, numerical_gradient, atol=1e-5
)
```
