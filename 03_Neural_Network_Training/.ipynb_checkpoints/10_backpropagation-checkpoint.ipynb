```python
import numpy as np

def sigmoid(z):
    return 1 / (1 + np.exp(-z))
```

```python


def forward(X, W, b):
    z = X @ W + b
    a = sigmoid(z)

    return z, a


def backward(X, y, W, b):
    n = len(X)

    # Forward pass
    z, a = forward(X, W, b)

    # Backward pass: BCE + sigmoid
    dz = (a - y) / n

    dW = X.T @ dz
    db = np.sum(dz, axis=0, keepdims=True)

    return a, dW, db
```

```python


X = np.array([
    [0, 0],
    [0, 1],
    [1, 0],
    [1, 1]
], dtype=float)

y = np.array([
    [0],
    [0],
    [0],
    [1]
], dtype=float)

W = np.zeros((2, 1))
b = np.zeros((1, 1))
```

```python

predictions, dW, db = backward(X, y, W, b)

print("Predictions before update:\n", predictions)
print("\nWeight gradients:\n", dW)
print("Bias gradient:\n", db)
```

```python
# Update parameters using gradients
learning_rate = 0.1

W = W - learning_rate * dW
b = b - learning_rate * db
```

```python
print("\nUpdated weights:\n", W)
print("Updated bias:\n", b)

_, new_predictions = forward(X, W, b)
print("Predictions after one update:\n", new_predictions)
```
