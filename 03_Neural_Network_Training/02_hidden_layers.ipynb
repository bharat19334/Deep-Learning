```python
import numpy as np
```

```python
def sigmoid(z):
    return 1 / (1 + np.exp(-z))
```

```python
X = np.array([[0.5, 0.8]])
```

```python
# Input layer to Hidden layer
W1 = np.array([
    [0.2, 0.4, 0.6],
    [0.3, 0.5, 0.7]
])
b1 = np.zeros((1, 3))

# Hidden layer to Output layer
W2 = np.array([
    [0.2],
    [0.4],
    [0.6]
])
b2 = np.zeros((1, 1))
```

```python
# Hidden layer
z1 = X @ W1 + b1
a1 = sigmoid(z1)

# Output layer
z2 = a1 @ W2 + b2
a2 = sigmoid(z2)

print("Input shape:", X.shape)
print("Hidden layer output:\n", a1)
print("Hidden layer shape:", a1.shape)
print("Final output:", a2)
```
