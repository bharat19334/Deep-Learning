```python
import numpy as np

np.random.seed(42)
```

```python
def sigmoid(z):
    z = np.clip(z, -500, 500)
    return 1 / (1 + np.exp(-z))

def binary_cross_entropy(y, predictions):
    predictions = np.clip(
        predictions, 1e-9, 1 - 1e-9
    )
    return -np.mean(
        y * np.log(predictions)
        + (1 - y) * np.log(1 - predictions)
    )
```

```python
# XOR 
X = np.array([
    [0, 0],
    [0, 1],
    [1, 0],
    [1, 1]
], dtype=float)

y = np.array([
    [0],
    [1],
    [1],
    [0]
], dtype=float)

# Network architecture: 2 inputs -> 4 hidden neurons -> 1 output
W1 = np.random.randn(2, 4)
b1 = np.zeros((1, 4))

W2 = np.random.randn(4, 1)
b2 = np.zeros((1, 1))

learning_rate = 1.0
epochs = 20000
n = len(X)
```

```python
for epoch in range(epochs):
    
    Z1 = X @ W1 + b1
    A1 = sigmoid(Z1)

    Z2 = A1 @ W2 + b2
    A2 = sigmoid(Z2)

    loss = binary_cross_entropy(y, A2)
    
    dZ2 = (A2 - y) / n
    dW2 = A1.T @ dZ2
    db2 = np.sum(dZ2, axis=0, keepdims=True)

    dA1 = dZ2 @ W2.T
    dZ1 = dA1 * A1 * (1 - A1)

    dW1 = X.T @ dZ1
    db1 = np.sum(dZ1, axis=0, keepdims=True)

    # parameter update
    W1 -= learning_rate * dW1
    b1 -= learning_rate * db1

    W2 -= learning_rate * dW2
    b2 -= learning_rate * db2

    if epoch % 5000 == 0:
        print(f"Epoch {epoch}: Loss = {loss:.6f}")
```

```python
# predictions
hidden_output = sigmoid(X @ W1 + b1)
final_output = sigmoid(hidden_output @ W2 + b2)

predictions = (final_output >= 0.5).astype(int)

print("\nActual values:", y.ravel())
print("Probabilities:", final_output.ravel())
print("Predictions:", predictions.ravel())

assert np.array_equal(predictions, y.astype(int))
```
