```python
import numpy as np
```

```python
def softmax(logits):
    logits = logits - np.max(logits, axis=1, keepdims=True)
    exp_values = np.exp(logits)

    return exp_values / np.sum(
        exp_values, axis=1, keepdims=True
    )
```

```python
def categorical_cross_entropy(y_true, probabilities):
    probabilities = np.clip(
        probabilities, 1e-9, 1.0
    )

    losses = -np.sum(
        y_true * np.log(probabilities), axis=1
    )

    return np.mean(losses)
```

```python
# Three samples and three classes
logits = np.array([
    [2.0, 1.0, 0.1],
    [0.5, 2.0, 0.3],
    [0.2, 0.4, 3.0]
])

# One-hot encoded labels
y_true = np.array([
    [1, 0, 0],
    [0, 1, 0],
    [0, 0, 1]
])
```

```python
probabilities = softmax(logits)
loss = categorical_cross_entropy(y_true, probabilities)

print("Probabilities:\n", probabilities)
print("Predicted classes:", np.argmax(probabilities, axis=1))
print("Categorical cross-entropy:", loss)
```
