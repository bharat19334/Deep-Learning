```python
import numpy as np

def sigmoid(z):
    return 1 / (1 + np.exp(-z)) 
```

```python
# AND gate 
X = np.array([
    [0, 0],
    [0, 1],
    [1, 0],
    [1, 1]
], dtype=float)
```

```python
# This is Single-layer network parameters
W = np.array([[10.0], [10.0]])
b = np.array([-15.0])
```

```python
# Forward pass
z = X @ W + b
output = sigmoid(z)
predictions = (output >= 0.5).astype(int)
```

```python
print("Weighted sums:\n", z)
print("Probabilities:\n", output)
print("Predictions:\n", predictions.ravel())
print("Expected: [0, 0, 0, 1]")
```
