```python
import numpy as np
```

```python
def sigmoid(z):
    return 1 / (1 + np.exp(-z))
    
def forward_propagation(X, W1, b1, W2, b2):
    z1 = X @ W1 + b1
    a1 = sigmoid(z1)

    z2 = a1 @ W2 + b2
    a2 = sigmoid(z2)

    return a1, a2
```

```python
X = np.array([
    [0, 0],
    [0, 1],
    [1, 0],
    [1, 1]
], dtype=float)

W1 = np.array([
    [0.2, 0.4, 0.6],
    [0.3, 0.5, 0.7]
])
b1 = np.zeros((1, 3))

W2 = np.array([
    [0.2],
    [0.4],
    [0.6]
])
b2 = np.zeros((1, 1))

hidden_output, final_output = forward_propagation(
    X, W1, b1, W2, b2
)
```

```python
print("Hidden layer:\n", hidden_output)
print("\nFinal output:\n", final_output)
```
