```python
import numpy as np

# Create a random generator with a fixed seed
rng = np.random.default_rng(42)

# Random integers
integers = rng.integers(0, 10, size=5)

# Random decimal values between 0 and 1
decimals = rng.random(5)

# Random values from a normal distribution
normal_values = rng.normal(loc=0, scale=1, size=5)

# Random matrix
matrix = rng.random((3, 3))
```

```python
print("Random integers:", integers)
print("Random decimals:", decimals)
print("Normal distribution:", normal_values)
print("\nRandom matrix:\n", matrix)
```

```python
# Same seed produces the same initial sequence
rng1 = np.random.default_rng(42)
rng2 = np.random.default_rng(42)

values1 = rng1.random(5)
values2 = rng2.random(5)

print("\nReproducible:", np.array_equal(values1, values2))
```
