```python
import numpy as np
arr = np.arange(1, 13)
```

```python
# Reshape into 2D
matrix = arr.reshape(3, 4)
print("Original matrix:\n", matrix)
```

```python
# Reshape into 3D
tensor = arr.reshape(2, 2, 3)
print("\n3D array:\n", tensor)
```

```python
# Automatic dimension calculation
reshaped = arr.reshape(3, -1)
print("\nReshaped array:\n", reshaped)
```

```python
# Transpose rows and columns
transposed = matrix.T
print("\nTranspose:\n", transposed)

print("Original shape:", matrix.shape)
print("Transpose shape:", transposed.shape)
```
