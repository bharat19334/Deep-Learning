```python
import numpy as np

a = np.array([1, 2, 3])
b = np.array([4, 5, 6])
```

```python
# Element-wise multiplication
element_wise = a * b
```

```python
# Dot product
dot_result = np.dot(a, b)

# Dot product using @
at_result = a @ b
```

```python
print("Vector A:", a)
print("Vector B:", b)
print("Element-wise multiplication:", element_wise)
print("Dot product:", dot_result)
print("Using @:", at_result)

assert dot_result == at_result
```
