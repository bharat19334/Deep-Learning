```python
import numpy as np

A = np.array([[1, 2],
              [3, 4]])

B = np.array([[5, 6],
              [7, 8]])
```

```python
# Element-wise multiplication
element_wise = A * B
```

```python
# Matrix multiplication
result = A @ B

# Alternative method
result_dot = np.matmul(A, B)
```

```python
print("Matrix A:\n", A)
print("\nMatrix B:\n", B)
print("\nElement-wise multiplication:\n", element_wise)
print("\nMatrix multiplication:\n", result)

assert np.array_equal(result, result_dot)
```

```python
# Matrix multiplication requires:
# A.shape[1] == B.shape[0]
print("\nA shape:", A.shape)
print("B shape:", B.shape)
print("Result shape:", result.shape)
```
