```python
# Variables and data types
name = "Bharat"
age = 21
cgpa = 8.0
is_student = True

print(type(name))
print(type(age))
print(type(cgpa))
print(type(is_student))
```

```python
# Lists
numbers = [10, 20, 30, 40, 50]

print(numbers[0])
print(numbers[-1])
print(numbers[1:4])
```

```python
# Loops
for number in numbers:
    print(number)
```

```python
# List comprehension
squares = [number ** 2 for number in numbers]
print(squares)
```

```python
# Functions
def calculate_mean(values):
    return sum(values) / len(values)
print(calculate_mean(numbers))
```

```python
# Dictionary
student = {
    "name": "Bharat",
    "subject": "Deep Learning"
}

print(student["name"])
print(student["subject"])
```
