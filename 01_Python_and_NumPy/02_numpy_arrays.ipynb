```python
import numpy as np

# Create arrays
arr = np.array([10, 20, 30, 40])
matrix = np.array([[1, 2, 3],
                   [4, 5, 6]])
```

```python
zeros = np.zeros((2, 3))
ones = np.ones((2, 3))
sequence = np.arange(0, 10, 2)
values = np.linspace(0, 1, 5)
```

```python
print("Array:", arr)
print("Matrix:\n", matrix)
print("Zeros:\n", zeros)
print("Ones:\n", ones)
print("Sequence:", sequence)
print("Linspace:", values)
```
