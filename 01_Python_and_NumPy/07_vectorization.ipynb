```python
import numpy as np
arr = np.arange(1, 6)
```

```python
# Using a loop
loop_result = []
for number in arr:
    loop_result.append(number ** 2)
    
loop_result = np.array(loop_result)
```

```python
# Using vectorization
vectorized_result = arr ** 2

print("Original:", arr)
print("Loop result:", loop_result)
print("Vectorized result:", vectorized_result)
```

```python
# Both methods should give the same result
assert np.array_equal(loop_result, vectorized_result)

# More vectorized operations
print("Add 5:", arr + 5)
print("Multiply by 2:", arr * 2)
print("Mean:", np.mean(arr))
print("Sum:", np.sum(arr))
```
